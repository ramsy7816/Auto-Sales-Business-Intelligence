**Auto Sales Business Insights**
## Project Objective
The objective of this analysis was to evluate sales perfomance, product demand, customer contribution, geographic perfomance, order activity, pricing patterns and sales trends using Python, SQL and Power BI.

## Key insights
1. Sales grew strongly in 2019 but declined sharply in 2020. 2019 was the strongest year in the dataset, with sales increasing by aproximately 39.3% compared to 2018. However, sales fell by approximately 62.8% in 2020


The business should investigate what drove the strong 2019 perfomance and what factors contributed to the substantial 2020 decline.


2. Classic Cars are the dorminant product line. classic cars generated approximately 39,4% of total sales, making it the largest revenue generating product line by a sunstantial margin.


3. The business has significant geographic concentration, with the USA alone contributing about a third of the total sales and the top three countries, USA, Spain, France, accounting for more than half of the revenue.


4. Medium sized deals generated approximately 60.8% of the total sales, making them the dorminant deal category.


5. shipped orders account for approximately 92.4% of the recorded sales, indicating that most sales value in the dataset is associated with shipped orders.


6. November stands out as an exceptionally strong sales month in both 2018 and 2019, with sales exceeding 1 million dolars in each year.
