```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
##import seaborn as sns

sales = pd.read_csv("Auto Sales data.csv")
```

```python
##convert the date into a datetime series
sales["ORDERDATE"] = pd.to_datetime(sales["ORDERDATE"], dayfirst = True)
```

```python
##get the information about the dataset
sales.info()
```

```python
##check for missing values
sales.isnull().sum()
```

```python
##calculate the percentage of missing values in the dataset
percent = (sales["ORDERDATE"].isnull().sum()/len(sales))*100
percent
```

```python
##check for duplicate rows
sales.duplicated().sum()
```

```python
##number of un unique items in each columns
categorical_columns = sales.select_dtypes(include ="object").columns
for column in categorical_columns:
    print(f"\n---{column}---")
    print("Unique Values: ",sales[column].nunique())
    print(sales[column].value_counts().head())
```

```python
##number of records per year
sales["ORDERDATE"].dt.year.value_counts().sort_index
```

```python
##validate the data if it is correct
sales["CALCULATED_SALES"] = sales["QUANTITYORDERED"] * sales["PRICEEACH"]
sales["SALES_DIFFERENCE"]=sales["CALCULATED_SALES"]- sales["SALES"]
sales[
    ["QUANTITYORDERED",
     "PRICEEACH",
     "SALES",
     "CALCULATED_SALES",
     "SALES_DIFFERENCE"]
].head(10)
```

```python
##check for anomalies
sales[sales["MSRP"] <= 0]
```
